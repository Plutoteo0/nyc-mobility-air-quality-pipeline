```python
from pyspark.sql import functions as F
import com.microsoft.spark.fabric

dates = spark.read.synapsesql("wh_gold.dbo.dim_date")
years = dates.select(F.col("year")).distinct()
gdps = spark.table("lh_silver.dbo.gdp")

gdp_per_year = years.join(gdps, on="year", how="left")
gdp_per_year.write.mode("overwrite").synapsesql("wh_gold.dbo.dim_gdp")
```
