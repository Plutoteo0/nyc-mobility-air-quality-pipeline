```python
from pyspark.sql import functions as F

df = spark.table("lh_bronze.dbo.ecb_usd_eur")

df = df.select([F.col("TIME_PERIOD"), F.col("OBS_VALUE")])
df = df.withColumnsRenamed({
    "TIME_PERIOD": "date",
    "OBS_VALUE": "usd_per_eur",
})
df.write.format("delta").mode("overwrite").saveAsTable("lh_silver.dbo.fx_usd_eur")
print(spark.read.format("delta").table("lh_silver.dbo.fx_usd_eur").count())
```
