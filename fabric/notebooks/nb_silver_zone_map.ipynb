```python
from pyspark.sql import functions as F

m = spark.read.option("header", True).option("inferSchema", True).csv("Files/reference/zone_station_map.csv")
m.printSchema()

m.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("lh_silver.dbo.zone_station_map")

s = spark.table("lh_silver.dbo.zone_station_map")
print(s.count())
s.groupBy("parameter").agg(F.sum(F.col("is_far_station").cast("int"))).show()
```
