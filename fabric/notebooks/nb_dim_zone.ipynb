```python
import com.microsoft.spark.fabric

df = spark.read.csv("Files/zone/taxi_zone_lookup.csv", header=True, inferSchema=True)

dim_zone = (
    df.withColumnRenamed("LocationID", "zone_id")
      .withColumnRenamed("Borough", "borough")
      .withColumnRenamed("Zone", "zone_name")
)

dim_zone.write.format("delta").mode("overwrite").synapsesql("wh_gold.dbo.dim_zone")
```
